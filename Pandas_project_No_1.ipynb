```python
import pandas as pd
import numpy as np
```

**Usecase-1 Load all three csv datasets in Pandas Data frames and display first 5 records.**

```python
df_customers = pd.read_csv("customers.csv")
df_products = pd.read_csv("products.csv")
df_transactions = pd.read_csv("transactions.csv")

print("Customers Data:")
display(df_customers.head())

print("Products Data:")
display(df_products.head())

print("Transactions Data:")
display(df_transactions.head())
```

**Usecase-2 Display only those customers from CSV_s1, who purchased more than 3 products**

```python
purchase_counts = df_transactions.groupby("'custno'").size()
more_than_3 = purchase_counts[purchase_counts > 3].index
```

```python
frequent_customers = df_customers[df_customers["'custno'"].isin(more_than_3)]
display(frequent_customers)
```

**Usecase-3 Display top 5 most demanded products from CSV_s1**

```python
product_counts = df_transactions["'productno'"].value_counts()
```

```python
top_5_products = product_counts.head(5)
print(top_5_products)
```

```python
top_5_df = top_5_products.reset_index()
top_5_df.columns = ["'productno'", "'purchase_count'"]

top_5_details = pd.merge(top_5_df, df_products, on="'productno'", how='left')
display(top_5_details)
```

**Usecase-4 Display top 5 transactions amount from CSV_s1**

```python
top_5_transactions = df_transactions.sort_values(by="'amount'", ascending=False).head(5)
display(top_5_transactions)
```
